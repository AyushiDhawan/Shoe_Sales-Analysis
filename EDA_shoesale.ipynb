# Identify outliers using boxplots or IQR method.

```python
#Boxplot for Price USD
sns.boxplot(x=df['Price_USD'], color='skyblue')
plt.title('Boxplot of Price USD - Outlier Detection')
plt.xlabel('Price (USD)')
plt.show()
```

```python
# Boxplot for Units Sold
sns.boxplot(x=df['Units_Sold'], color='skyblue')
plt.title('Boxplot of Units Sold - Outlier Detection')
plt.xlabel('Units Sold')
plt.show()
```

```python
# Boxplot for Revenue USD
sns.boxplot(x=df['Revenue_USD'], color='skyblue')
plt.title('Boxplot of Revenue USD - Outlier Detection')
plt.xlabel('Revenue (USD)')
plt.show()
```

```python
q1= df["Revenue_USD"].quantile(0.25)
q3 = df["Revenue_USD"].quantile(0.75)
IQR = q3 - q1
lower_bound = q1- 1.5 * IQR
upper_bound = q3 + 1.5 * IQR
outliers = df[(df["Revenue_USD"] < lower_bound) | (df["Revenue_USD"] > upper_bound)]
outliers
```

# Create correlation matrix to find relationships between variables.

```python
# Correlation matrix 
correlation_matrix = df[['Price_USD', 'Units_Sold', 'Revenue_USD']].corr() 
print(correlation_matrix) 
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, square=True) 
plt.title('Correlation Matrix') 
plt.show()
```

# Summarize key insights and patterns for business understanding.

Revenue depends more on selling more shoes than increasing prices.

New Balance gives the highest average revenue, while Skechers has the most sales transactions.

Sneakers and Boots perform better than Sports and Casual shoes.

UAE and USA generate higher revenue, while India has lower average sales.

Sales are balanced across Retail Stores, Online, and Mall channels, with Retail Stores slightly leading.

The dataset is clean with no missing values or duplicates.


A few high revenue outliers exist, but they appear to be genuine large purchases.
